```python
import torch
from transformers import AutoModelForSpeechSeq2Seq, AutoProcessor, pipeline
# Use Metal Performance Shaders (MPS) backend Mac's GPU - Metal
device = "mps"
torch_dtype = torch.float16
# You can replace the model on the openAI's (HuggingFace Hub)[https://huggingface.co/openai]
# Ex. "openai/whisper-large-v3-turbo" can reduce the inference time by 80%, at the expense of a minor quality degradation like openai/whisper-medium-v3.
model_id = "openai/whisper-large-v3"


model = AutoModelForSpeechSeq2Seq.from_pretrained(
    model_id, torch_dtype=torch_dtype, low_cpu_mem_usage=True, use_safetensors=True
)
model.to(device)
```

```python
processor = AutoProcessor.from_pretrained(model_id)
pipe = pipeline(
    "automatic-speech-recognition",
    model=model,
    tokenizer=processor.tokenizer,
    feature_extractor=processor.feature_extractor,
    torch_dtype=torch_dtype,
    device=device,
    decoder_kwargs={
        "skip_special_tokens": True
    }
)
```

```python
pipe.model.config.forced_decoder_ids = (
  pipe.tokenizer.get_decoder_prompt_ids(
    language="chinese",
    task="transcribe"
  )
)
```

```python
prompt_text = "這是一段以藏文跟正體中文講解的佛法研討班錄音"
prompt_ids = pipe.tokenizer.get_prompt_ids(prompt_text, return_tensors="pt").to(device)
result = pipe("test.mp3",
              return_timestamps=True,
              generate_kwargs={
                  "prompt_ids": prompt_ids
              })
print(result)
```
